```python
import numpy as np
import scipy 
from scipy import stats 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
size=1000000
fig,ax=plt.subplots(3,3,figsize=(12,6),sharey=False,)
x0=np.random.normal(loc=100,scale=15,size=size)

sns.kdeplot(x0,ax=ax[0][0])
ax[0][0].axvline(115,0,0.57,lw=2,ls="--",color="red")
x_f1=np.linspace(0,115,100)
y_f1=stats.norm.pdf(x_f1,loc=100,scale=15)
ax[0][0].fill_between(x_f1,y_f1,color="teal")



sns.kdeplot(x0,ax=ax[0][1])
ax[0][1].axvline(131.5,0,0.11,lw=2,ls="--",color="red")
x_f2=np.linspace(0,132.5,100)
y_f2=stats.norm.pdf(x_f2,loc=100,scale=15)
ax[0][1].fill_between(x_f2,y_f2,color="cyan")

sns.kdeplot(x0,ax=ax[0][2])
ax[0][2].axvline(90.5,0,0.78,lw=2,ls="--",color="red")
ax[0][2].axvline(110.5,0,0.75,lw=2,ls="--",color="red")
x_f3=np.linspace(90,111,100)
y_f3=stats.norm.pdf(x_f3,loc=100,scale=15)
ax[0][2].fill_between(x_f3,y_f3,color="beige")


#----------------------------------------------------------
x1=np.random.normal(loc=98.2,scale=0.62,size=size)
x2=np.random.normal(loc=998,scale=202,size=size)
#----------------------------------------------------------

sns.kdeplot(x1,ax=ax[1][0],color="cyan")
ax[1][0].axvline(100.6,0,0.05,lw=5,ls="--",color="red")
x_f4=np.linspace(95,101.6,100)
y_f4=stats.norm.pdf(x_f4,loc=98.2,scale=0.62)
ax[1][0].fill_between(x_f4,y_f4,color="yellow")

sns.kdeplot(x2,ax=ax[1][1],color="indigo")
ax[1][1].axvline(1100,0,0.8,lw=2,ls="--",color="red")
x_f5=np.linspace(0,1100,100)
y_f5=stats.norm.pdf(x_f5,loc=998,scale=202)
ax[1][1].fill_between(x_f5,y_f5,color="purple")

sns.kdeplot(x2,ax=ax[1][2],color="indigo")
ax[1][2].axvline(1040,0,0.93,lw=2,ls="--",color="red")
x_f6=np.linspace(0,1040,100)
y_f6=stats.norm.pdf(x_f6,loc=998,scale=202)
ax[1][2].fill_between(x_f6,y_f6,color="black")


x3=np.random.normal(loc=8.2,scale=1.1,size=size)
sns.kdeplot(x3,ax=ax[2][0],color="green")
ax[2][0].axvline(5,0,0.05,lw=5,ls="--",color="red")
x_f7=np.linspace(0,5,100)
y_f7=stats.norm.pdf(x_f7,loc=8.2,scale=1.1)
ax[2][0].fill_between(x_f7,y_f7,color="teal")

sns.kdeplot(x3,ax=ax[2][1],color="green")
ax[2][1].axvline(5.67,0,0.1,lw=5,ls="--",color="red")
x_f8=np.linspace(0,5.67,100)
y_f8=stats.norm.pdf(x_f8,loc=8.2,scale=1.1)
ax[2][1].fill_between(x_f8,y_f8,color="pink")

sns.kdeplot(x3,ax=ax[2][2],color="green")
# fig.tight_layout()

# for i in range(3):
#     for j in range(3):
#         ax[i, j].axis("off")

# for i in range(3):
#     for j in range(3):
#         ax[i, j].get_yaxis().set_visible(False)

# turn off the y lable x lable
for i in range(3):
    for j in range(3):
        ax[i, j].set_ylabel("")
        ax[i, j].set_xlabel("")

plt.suptitle("Gaussian Distribution Probability Quiz",size=20,color="red",y=0.95,x=0.5)
```
